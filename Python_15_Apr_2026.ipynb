```python
def factorial(n):
  fact = 1
  if n < 1:
    return "Negative Number not allowed"
  elif n == 0:
    return 1
  else:
    for i in range(1,n+1):
      fact = fact * i
  return fact
factorial(3)
```

```python
def check_palindrome(s):
    s = s.lower()          # fix case issue
    i = 0
    l = len(s) - 1       # fix index

    while i < l :
        if s[i] == s[l]:
            i += 1
            l -= 1
        else:
            return "Not Palindrome"

    return "Palindrome"

print(check_palindrome("Nayan"))
```

```python
# Remove Duplicates

l = [1,2,1,4,2,3,6,3,6,2,4,5,7,8,7,8]

def remove_duplicates(l):
  lst = []
  for i in range(0,len(l)):
    if l[i] not in lst:
      lst.append(l[i])
  return lst
print(remove_duplicates(l))
```

```python
# Two Sum Problem

arr = [2,11,7,15]
t = 9

def two_sum(arr, t):
    seen = {}

    for i, num in enumerate(arr):
        diff = t - num

        if diff in seen:
            return [seen[diff], i]

        seen[num] = i

    return None

print(two_sum(arr,t))
```

```python
import threading

def task(name):
    print(f"Task {name} started")
    for i in range(3):
        print(f"{name}: {i}")
    print(f"Task {name} finished")

# Create threads
t1 = threading.Thread(target=task, args=("A",))
t2 = threading.Thread(target=task, args=("B",))

# Start threads
t1.start()
t2.start()

# Wait for both to finish
t1.join()
t2.join()

print("All tasks done")
```

```python
import threading
import time

result = []

def task(n):
    time.sleep(0.1)
    result.append(n)

threads = []

for i in range(5):
    t = threading.Thread(target=task, args=(i,))
    t.start()
    threads.append(t)

for t in threads:
    t.join()

# No join here
print("Result:", result)
```

```python
from multiprocessing import Process
import time

def task(name):
    print(f"{name} started")
    for i in range(3):
        print(f"{name}: {i}")
        time.sleep(1)  # simulate work
    print(f"{name} finished")

if __name__ == "__main__":
    # Create processes
    p1 = Process(target=task, args=("Process A",))
    p2 = Process(target=task, args=("Process B",))

    # Start processes
    p1.start()
    p2.start()

    # Wait for processes to complete
    p1.join()
    p2.join()

    print("All processes completed")
```

```python
# Two Sum Problem multiple pairs
arr = [7, 11, 2, 15, 4, -2]
t = 9
def two_sum(arr, t):
  seen = {}
  res = []
  for i, v in enumerate(arr):
    diff = t - v
    if diff in seen:
      res.append([seen[diff], i])
    seen[v] = i
  return res

print(two_sum(arr,t))
```

```python
# Sliding Window: To find the max in the window size

# Brute Force Appraoch

arr = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3

def demo(arr, k):
  res = []
  for i in range(len(arr) - k + 1):
    max = arr[i]
    for j in range(i, i+k):
      if arr[j] > max:
        max = arr[j]
    res.append(max)
  return res

print(demo(arr,k))
```
